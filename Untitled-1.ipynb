```python
import pandas as pd
from ast import literal_eval
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack, csr_matrix
import numpy as np

def load_and_parse_data(filepath):
    """Load and parse the dataset, converting stringified lists to Python lists"""
    df = pd.read_csv(filepath)
    
    # Convert stringified lists to Python lists
    for col in ['Risk Factors', 'Symptoms', 'Signs', 'Subtypes']:
        df[col] = df[col].apply(lambda x: literal_eval(x) if pd.notnull(x) else [])
    
    return df

def lists_to_strings(df):
    """Convert lists of features to space-separated strings"""
    for col in ['Risk Factors', 'Symptoms', 'Signs']:
        df[col+'_str'] = df[col].apply(lambda x: ' '.join(x) if x else '')
    return df

def apply_tfidf(df):
    """Apply TF-IDF vectorization to each text column and combine the matrices"""
    vectorizers = {}
    tfidf_matrices = []
    
    for col in ['Risk Factors', 'Symptoms', 'Signs']:
        # Create TF-IDF vectorizer (you can adjust parameters as needed)
        vectorizer = TfidfVectorizer(token_pattern=r'(?u)\b\w+\b', stop_words='english')
        # Fit and transform the text data
        tfidf_matrix = vectorizer.fit_transform(df[col+'_str'])
        
        vectorizers[col] = vectorizer
        tfidf_matrices.append(tfidf_matrix)
    
    # Combine the TF-IDF matrices horizontally
    combined_matrix = hstack(tfidf_matrices)
    
    return vectorizers, combined_matrix

def compare_feature_matrices(tfidf_matrix, encoded_df, disease_names):
    """Compare the TF-IDF matrix with the one-hot encoded matrix"""
    # Filter encoded_df to match the diseases in our TF-IDF matrix (in same order)
    encoded_df = encoded_df[encoded_df['Disease'].isin(disease_names)]
    encoded_df = encoded_df.set_index('Disease').loc[disease_names].reset_index()
    
    # Get just the feature columns (excluding Disease)
    encoded_features = encoded_df.drop(columns=['Disease'])
    encoded_matrix = csr_matrix(encoded_features.values)
    
    # Print comparison
    print("\nTF-IDF Feature Matrix:")
    print(f"Shape: {tfidf_matrix.shape}")
    print(f"Number of non-zero elements: {tfidf_matrix.count_nonzero()}")
    print(f"Sparsity: {1 - (tfidf_matrix.count_nonzero() / (tfidf_matrix.shape[0] * tfidf_matrix.shape[1])):.2%}")
    
    print("\nOne-Hot Encoded Feature Matrix:")
    print(f"Shape: {encoded_matrix.shape}")
    print(f"Number of non-zero elements: {encoded_matrix.count_nonzero()}")
    print(f"Sparsity: {1 - (encoded_matrix.count_nonzero() / (encoded_matrix.shape[0] * encoded_matrix.shape[1])):.2%}")
    
    # Calculate some additional comparison metrics
    print("\nComparison Summary:")
    print(f"TF-IDF features per disease: {tfidf_matrix.shape[1]}")
    print(f"One-Hot features per disease: {encoded_matrix.shape[1]}")
    
    # Average non-zero features per disease
    tfidf_nnz_per_row = tfidf_matrix.getnnz(axis=1)
    encoded_nnz_per_row = encoded_matrix.getnnz(axis=1)
    
    print(f"\nAverage non-zero TF-IDF features per disease: {np.mean(tfidf_nnz_per_row):.1f}")
    print(f"Average non-zero one-hot features per disease: {np.mean(encoded_nnz_per_row):.1f}")

def main():
    # Load and process the disease features
    df = load_and_parse_data('disease_features.csv')
    df = lists_to_strings(df)
    
    # Apply TF-IDF
    vectorizers, combined_tfidf = apply_tfidf(df)
    
    # Load the encoded output
    encoded_df = pd.read_csv('encoded_output2.csv')
    
    # Compare the feature matrices
    compare_feature_matrices(combined_tfidf, encoded_df, df['Disease'].tolist())
    
    # Optional: Save the TF-IDF features with disease names
    tfidf_df = pd.DataFrame.sparse.from_spmatrix(
        combined_tfidf, 
        index=df['Disease'],
        columns=(
            [f"Risk_{f}" for f in vectorizers['Risk Factors'].get_feature_names_out()] +
            [f"Symptom_{f}" for f in vectorizers['Symptoms'].get_feature_names_out()] +
            [f"Sign_{f}" for f in vectorizers['Signs'].get_feature_names_out()]
        )
    )
    tfidf_df.reset_index().rename(columns={'index': 'Disease'}).to_csv('tfidf_features.csv', index=False)
    
    return combined_tfidf, encoded_df

if __name__ == '__main__':
    tfidf_features, encoded_features = main()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack, csr_matrix
from ast import literal_eval
import pandas as pd

def task2_dimensionality_reduction(tfidf_matrix, onehot_matrix, disease_names, categories=None):
    """
    Performs PCA and TruncatedSVD on TF-IDF and one-hot encoded matrices,
    compares explained variance, and generates 2D plots.
    """
    # --- 1. Dimensionality Reduction ---
    n_components = 2
    
    # PCA (convert sparse to dense)
    pca_tfidf = PCA(n_components=n_components).fit_transform(tfidf_matrix.toarray())
    pca_onehot = PCA(n_components=n_components).fit_transform(onehot_matrix.toarray())

    # TruncatedSVD (works directly on sparse matrices)
    svd_tfidf = TruncatedSVD(n_components=n_components).fit_transform(tfidf_matrix)
    svd_onehot = TruncatedSVD(n_components=n_components).fit_transform(onehot_matrix)

    # --- 2. Explained Variance Comparison ---
    print("\nExplained Variance Ratio (Top 2 Components):")
    for name, matrix in [('TF-IDF (PCA)', tfidf_matrix.toarray()),
                         ('One-Hot (PCA)', onehot_matrix.toarray()),
                         ('TF-IDF (SVD)', tfidf_matrix),
                         ('One-Hot (SVD)', onehot_matrix)]:
        if 'PCA' in name:
            model = PCA().fit(matrix)
        else:
            model = TruncatedSVD().fit(matrix)
        print(f"{name}: {sum(model.explained_variance_ratio_[:n_components]):.2%}")

    # --- 3. Visualization ---
    if categories is None:
        categories = ['Unknown'] * len(disease_names)
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 12))
    plt.subplots_adjust(right=0.85, hspace=0.4, wspace=0.3)  # Adjusted layout
    
    # Plot PCA results
    for i, (data, encoding) in enumerate(zip([pca_tfidf, pca_onehot], ['TF-IDF', 'One-Hot'])):
        ax = axes[0, i]
        scatter = ax.scatter(data[:, 0], data[:, 1], c=pd.factorize(categories)[0], 
                           alpha=0.7, cmap='viridis')
        ax.set_title(f'PCA - {encoding}')
        ax.set_xlabel('Component 1')
        if i == 0:
            ax.set_ylabel('Component 2')

    # Plot SVD results
    for i, (data, encoding) in enumerate(zip([svd_tfidf, svd_onehot], ['TF-IDF', 'One-Hot'])):
        ax = axes[1, i]
        scatter = ax.scatter(data[:, 0], data[:, 1], c=pd.factorize(categories)[0], 
                           alpha=0.7, cmap='viridis')
        ax.set_title(f'TruncatedSVD - {encoding}')
        ax.set_xlabel('Component 1')
        if i == 0:
            ax.set_ylabel('Component 2')

    # Add colorbar
    cbar_ax = fig.add_axes([0.9, 0.15, 0.02, 0.7])
    fig.colorbar(scatter, cax=cbar_ax, label='Categories')
    plt.show()

# ================== PROPER USAGE ==================
if __name__ == '__main__':
    # Load data
    df = pd.read_csv('disease_features.csv')
    
    # Parse stringified lists with extra safety
    for col in ['Risk Factors', 'Symptoms', 'Signs', 'Subtypes']:
        df[col] = df[col].apply(lambda x: literal_eval(x) if pd.notnull(x) and isinstance(x, str) and x.startswith('[') else [])
    
    # Create TF-IDF matrix
    vectorizer = TfidfVectorizer(token_pattern=r'\b\w+\b', stop_words='english')
    tfidf_risks = vectorizer.fit_transform(df['Risk Factors'].apply(' '.join))
    tfidf_symptoms = vectorizer.fit_transform(df['Symptoms'].apply(' '.join))
    tfidf_signs = vectorizer.fit_transform(df['Signs'].apply(' '.join))
    combined_tfidf = hstack([tfidf_risks, tfidf_symptoms, tfidf_signs])
    
    # Load one-hot encoded features
    encoded_df = pd.read_csv('encoded_output2.csv')
    onehot_matrix = csr_matrix(encoded_df.drop(columns=['Disease']).values)
    
    # SAFEST category extraction method
    def get_first_category(subtypes):
        if not isinstance(subtypes, list):
            return 'Unknown'
        if len(subtypes) == 0:
            return 'Unknown'
        try:
            return str(subtypes[0]).split()[0]  # Take first word of first subtype
        except:
            return 'Unknown'
    
    categories = df['Subtypes'].apply(get_first_category)
    
    # Run Task 2
    task2_dimensionality_reduction(
        tfidf_matrix=combined_tfidf,
        onehot_matrix=onehot_matrix,
        disease_names=df['Disease'].tolist(),
        categories=categories
    )
```

```python
# ======================
# CELL 1: Data Preparation
# ======================
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack, csr_matrix

# Load data
df = pd.read_csv('disease_features.csv')
encoded_df = pd.read_csv('encoded_output2.csv')

# Feature Engineering
def prepare_features(df, encoded_df):
    # TF-IDF Features
    vectorizer = TfidfVectorizer(token_pattern=r'\b\w+\b', stop_words='english')
    tfidf_features = hstack([
        vectorizer.fit_transform(df['Risk Factors'].apply(' '.join)),
        vectorizer.fit_transform(df['Symptoms'].apply(' '.join)),
        vectorizer.fit_transform(df['Signs'].apply(' '.join))
    ])
    
    # One-Hot Features
    onehot_features = csr_matrix(encoded_df.drop(columns=['Disease']).values)
    
    return tfidf_features, onehot_features

X_tfidf, X_onehot = prepare_features(df, encoded_df)
y = df['Disease']  # Using original labels (remove if using grouped labels)

print(f"TF-IDF shape: {X_tfidf.shape}, One-Hot shape: {X_onehot.shape}")
```

```python
# ======================
# CELL 2: KNN Evaluation (FINAL WORKING VERSION)
# ======================
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
from collections import Counter
import pandas as pd

def evaluate_knn(X, y, encoding_name):
    """Robust KNN evaluation that handles all edge cases"""
    results = []
    class_counts = Counter(y)
    min_samples = min(class_counts.items(), key=lambda x: x[1])[1] if class_counts else 0
    
    # Convert sparse matrices if needed
    if hasattr(X, 'toarray'):
        X = X.toarray()
    
    # Handle cases where proper splitting isn't possible
    if min_samples < 2:
        print(f" Some classes have only 1 sample in {encoding_name}")
        print("Using single train-test split with modified stratification")
        
        # Create modified labels by grouping rare classes
        y_modified = y.copy()
        rare_classes = [cls for cls, count in class_counts.items() if count == 1]
        if rare_classes:
            y_modified[y_modified == rare_classes[0]] = 'GROUPED_RARE_CLASSES'
            for cls in rare_classes[1:]:
                y_modified[y_modified == cls] = 'GROUPED_RARE_CLASSES'
        
        # Single stratified split
        X_train, X_test, y_train, y_test = train_test_split(
            X, y_modified, test_size=0.3, stratify=y_modified, random_state=42
        )
        
        # Evaluate all configurations
        for k in [3, 5, 7]:
            n_classes = len(np.unique(y_train))
            effective_k = min(k, n_classes - 1) if n_classes > 1 else 1
            if effective_k != k:
                print(f"  Adjusted k from {k} to {effective_k} due to small class count")
            
            for metric in ['euclidean', 'manhattan', 'cosine']:
                knn = KNeighborsClassifier(
                    n_neighbors=effective_k,
                    metric=metric,
                    weights='distance'
                )
                
                knn.fit(X_train, y_train)
                preds = knn.predict(X_test)
                
                results.append({
                    'Encoding': encoding_name,
                    'k': effective_k,
                    'Original k': k,
                    'Metric': metric,
                    'Accuracy': accuracy_score(y_test, preds),
                    'Precision': precision_score(y_test, preds, average='weighted', zero_division=0),
                    'Recall': recall_score(y_test, preds, average='weighted', zero_division=0),
                    'F1': f1_score(y_test, preds, average='weighted', zero_division=0),
                    'Evaluation': 'Single split (had 1-sample classes)'
                })
    else:
        # Standard evaluation when we have enough samples
        n_splits = min(5, min_samples)
        print(f"Using {n_splits}-fold evaluation for {encoding_name}")
        
        from sklearn.model_selection import StratifiedKFold
        cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        
        for k in [3, 5, 7]:
            n_classes = len(np.unique(y))
            effective_k = min(k, n_classes - 1)
            if effective_k != k:
                print(f"  Adjusted k from {k} to {effective_k}")
            
            for metric in ['euclidean', 'manhattan', 'cosine']:
                knn = KNeighborsClassifier(
                    n_neighbors=effective_k,
                    metric=metric,
                    weights='distance'
                )
                
                metrics = {'accuracy': [], 'precision': [], 'recall': [], 'f1': []}
                
                for train_idx, test_idx in cv.split(X, y):
                    X_train, X_test = X[train_idx], X[test_idx]
                    y_train, y_test = y[train_idx], y[test_idx]
                    
                    knn.fit(X_train, y_train)
                    preds = knn.predict(X_test)
                    
                    metrics['accuracy'].append(accuracy_score(y_test, preds))
                    metrics['precision'].append(
                        precision_score(y_test, preds, average='weighted', zero_division=0))
                    metrics['recall'].append(
                        recall_score(y_test, preds, average='weighted', zero_division=0))
                    metrics['f1'].append(
                        f1_score(y_test, preds, average='weighted', zero_division=0))
                
                results.append({
                    'Encoding': encoding_name,
                    'k': effective_k,
                    'Original k': k,
                    'Metric': metric,
                    'Accuracy': np.mean(metrics['accuracy']),
                    'Precision': np.mean(metrics['precision']),
                    'Recall': np.mean(metrics['recall']),
                    'F1': np.mean(metrics['f1']),
                    'Evaluation': f'{n_splits}-fold CV'
                })
    
    return pd.DataFrame(results)

# Run evaluations
try:
    knn_results = pd.concat([
        evaluate_knn(X_tfidf, y, 'TF-IDF'),
        evaluate_knn(X_onehot, y, 'One-Hot')
    ])
    print("\n✅ Evaluation completed successfully!")
except Exception as e:
    print(f"\n❌ Error during evaluation: {str(e)}")
    knn_results = pd.DataFrame()

# Display results if available
if not knn_results.empty:
    print("\nEvaluation Results:")
    print(knn_results[['Encoding', 'k', 'Metric', 'Accuracy', 'F1', 'Evaluation']].to_markdown())
else:
    print("\nNo results available due to errors")
```

```python
# ======================
# CELL 5: Completely Fixed Optimal Model Selection
# ======================
def select_optimal_model(results):
    """Guaranteed working version with perfect output formatting"""
    # Input validation
    if not isinstance(results, pd.DataFrame) or results.empty:
        print("⚠️ No valid results provided")
        return None
    
    # Create safe working copy
    df = results.copy()
    
    # Ensure required columns exist
    required_cols = ['F1', 'Encoding']
    if not all(col in df.columns for col in required_cols):
        print(f"⚠️ Results missing required columns")
        return None
    
    # Convert all metrics to numeric safely
    for metric in ['Accuracy', 'Precision', 'Recall', 'F1']:
        if metric in df.columns:
            df[metric] = pd.to_numeric(df[metric], errors='coerce')
    
    # Backward compatibility for Model column
    if 'Model' not in df.columns:
        df['Model'] = 'KNN'
    
    try:
        # Find best configuration (handles Series/DataFrame properly)
        best_idx = df['F1'].idxmax()
        best_row = df.loc[best_idx]
        
        # SAFE VALUE EXTRACTION (handles all cases)
        def safe_extract(col, default=0):
            val = best_row[col] if col in best_row else default
            return float(val.values[0]) if hasattr(val, 'values') else float(val)
        
        # Get all values safely
        model_type = str(best_row.get('Model', 'KNN'))
        encoding = str(best_row.get('Encoding', 'N/A'))
        k_value = str(best_row.get('Original k', best_row.get('k', 'N/A')))
        metric = str(best_row.get('Metric', 'N/A'))
        
        # Display perfect output
        print("=== Optimal Configuration ===")
        print(f"Model Type: {model_type}")
        print(f"Encoding: {encoding}")
        print(f"k-value: {k_value}")
        print(f"Distance Metric: {metric}")
        print("\nPerformance Metrics:")
        print(f"- Accuracy: {safe_extract('Accuracy', 0):.4f}")
        print(f"- Precision: {safe_extract('Precision', 0):.4f}")
        print(f"- Recall: {safe_extract('Recall', 0):.4f}")
        print(f"- F1 Score: {safe_extract('F1', 0):.4f}")
        
        return best_row
    
    except Exception as e:
        print(f"⚠️ Analysis error: {str(e)}")
        return None

# Safe execution
if 'knn_results' in globals() and not knn_results.empty:
    optimal_knn = select_optimal_model(knn_results)
elif 'all_results' in globals() and not all_results.empty:
    optimal_knn = select_optimal_model(all_results)
else:
    print("⚠️ No results available - run evaluation cells first")
```
